```python
sales["Shipping_Cost"].value_counts().head(10)
```

**Observation (negative totals):**
- 5 orders (0.002%) have a negative Total_Amount. All used DIWALI100, a flat ₹100 coupon, on items costing ₹26–36.
- 107 orders have a coupon discount larger than the order value; in most, shipping charges kept the total above zero.
- Manual check confirms: Total_Amount = Order_Value + Shipping_Cost − Coupon_Discount.
- Business insight: flat coupons appear to have no minimum order value, creating loss-making orders.
- Cleaning decision: cap Total_Amount at 0 and add a flag column (a checkout never charges below ₹0; revenue impact ≈ ₹63).

```python
paid_shipping = sales["Shipping_Cost"] > 0

print("Orders that paid shipping:", paid_shipping.sum())
print("Highest order value that PAID shipping:", sales.loc[paid_shipping, "Order_Value"].max())
print("Lowest order value with FREE shipping:", sales.loc[~paid_shipping, "Order_Value"].min())
```

- Free-shipping threshold test: every order below ₹499.34 paid shipping, and every order from ₹500.22 upward shipped free. This shows a clear business rule: free shipping on orders above ~₹500. About 7% of orders (17,541) fall below it.

## 8. How the Tables Connect (Key Checks)

```python
print("Order_ID unique in sales:", sales["Order_ID"].is_unique)
print("Customer_ID unique in customers:", customers["Customer_ID"].is_unique)
print("Product_ID unique in products:", products["Product_ID"].is_unique)
```

```python
missing_customers = ~sales["Customer_ID"].isin(customers["Customer_ID"])
missing_products = ~sales["Product_ID"].isin(products["Product_ID"])

print("Sales with unknown customer:", missing_customers.sum())
print("Sales with unknown product:", missing_products.sum())
```

```python
sales["Order_Date"].head(3)
```

**Observation (key checks):**
- Order_ID is unique, so each row represents one order.
- Customer_ID and Product_ID are unique in their lookup tables, so joins will not duplicate rows.
- No orphan records: every sale matches an existing customer and product. The tables can be joined safely.

## Summary of Data Understanding

**Data overview**
- sales: 250,000 orders (one row = one order), customers: 40,000, products: 2,000.
- Period: 1 June 2024 – 30 June 2026 (25 months, including two Diwali seasons).
- Coverage: 10 states and 39 cities only.

**Data quality**
- No duplicate rows. All IDs are unique, and there are no orphan records, so the tables can be joined safely.
- Money formulas verified: Order_Value = Quantity × Unit_Price; Total_Amount = Order_Value + Shipping_Cost − Coupon_Discount.
- Coupon_Code blanks (80%) = no coupon used (verified: discount is 0 in all of them).
- Rating/Review blanks (52%) are structural: only delivered orders are rated; ~60% of delivered orders have a rating.
- 5 orders have a negative Total_Amount, caused by flat coupons on items under ₹40.

**Business rules discovered**
- Free shipping on orders above ~₹500 (no overlap between paid and free shipping around this value).
- Flat coupons (DIWALI100, FLAT50) have no minimum order value, creating loss-making orders.

**Metric definitions**
- GMV = Total_Amount of all orders. Net Revenue = Total_Amount of Delivered orders only.
- Processing/Shipped orders are treated as "Incomplete".
